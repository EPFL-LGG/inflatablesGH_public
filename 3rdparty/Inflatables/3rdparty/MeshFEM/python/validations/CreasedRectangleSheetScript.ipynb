```python
# low alpha test
alpha = np.pi
thickness = 5e-3
mA = 1e-3
gra = -9.80635e-3
sm,creasesd = semisphere_convergence.getCreasedRecSheetMesh(alpha,maxArea=mA)
es = semisphere_convergence.getElasticSheet(sm,thickness,creases=creasesd)
```

```python
mV = 2e-3
tm = semisphere_convergence.getCreasedRecTetMesh(alpha,thickness,maxVol=mV)
esolid = semisphere_convergence.getElasticSolid(tm)
```

```python
esim, tt = semisphere_convergence.gravitySimulation(esolid,gravity=gra)
esim_energy = esim.energy()
print(esim_energy)
```

```python
import tri_mesh_viewer
tet_viewer = tri_mesh_viewer.Viewer(esim, wireframe=True)
tet_viewer.show()
```
