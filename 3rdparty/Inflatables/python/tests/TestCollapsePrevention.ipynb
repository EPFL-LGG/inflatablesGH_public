```python
import sys
sys.path.append('..')
import metric_fit, parametrization, numpy as np, mesh, fd_validation
from matplotlib import pyplot as plt
from tri_mesh_viewer import TriMeshViewer
```

```python
cpe = metric_fit.CollapsePreventionEnergy()
cpe.activationThreshold = 0.2

def eval(x):
    cpe.setDet(x)
    return cpe.energy()
def eval_der(x):
    cpe.setDet(x)
    return cpe.denergy_ddet()
def eval_d2(x):
    cpe.setDet(x)
    return cpe.d2energy_d2det()

x = np.linspace(1e-1, 0.6, 100)
y = np.array(list(map(eval, x)))
dy = np.array(list(map(eval_der, x)))
d2y = np.array(list(map(eval_d2, x)))
```

```python
fig = plt.figure(figsize=(15,5))
plt.subplot(1, 3, 1)
plt.plot(x, y)
plt.subplot(1, 3, 2)
plt.plot(x, dy, 'o', x, np.gradient(y, x))
plt.subplot(1, 3, 3)
plt.plot(x, d2y, 'o', x, np.gradient(dy, x))
plt.show()
```
