```python
import sys
sys.path.append('..')
import metric_fit, parametrization, numpy as np, mesh
from grid_gen import triangulatedGrid2D
from tri_mesh_viewer import TriMeshViewer
```

```python
nx, ny = 30, 60
V, F = triangulatedGrid2D(nx, ny, triangulationType=2)
firstStripVertices = len(V)-np.arange(2 * nx) - 1

fitter = metric_fit.MetricFitter(metric_fit.Mesh2D(V, F))
fitter.setVars(fitter.getIdentityImmersion().transpose().ravel())
```

```python
viewer = TriMeshViewer(mesh.Mesh(fitter.getImmersion().transpose(), F))
viewer.showWireframe(True)
viewer.show()
```

```python
fitter.bendingStiffness = 1e-3
fitter.gravityVector = [0, 0, -0.02]

firstStripDoFs = np.array([[3 * v + 0, 3 * v + 1, 3 * v + 2] for v in firstStripVertices]).ravel()
metric_fit.benchmark_reset()
metric_fit.fit_metric_newton(fitter, firstStripDoFs)
metric_fit.benchmark_report()
immersedSurface = mesh.Mesh(fitter.getImmersion().transpose(), F)
viewer.update(mesh=immersedSurface)
```
