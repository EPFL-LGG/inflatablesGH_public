```python
import parametrization, numpy as np
from matplotlib import pyplot as plt
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
xmin = parametrization.circularMean(angles)
print(np.rad2deg(xmin))
plt.plot(x, [parametrization.sumSquaredCircularDist(x_i, angles) for x_i in x])
plt.scatter([xmin], [parametrization.sumSquaredCircularDist(xmin, angles)])
plt.show()
```
