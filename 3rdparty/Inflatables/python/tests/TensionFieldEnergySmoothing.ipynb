```python
import sys
sys.path.append('..')
import inflation, numpy as np, importlib, fd_validation, mesh
```

```python
tfe = inflation.OptionalTensionFieldEnergy([[1, 0],[0, 1]])
```

```python
tfe.energy()
```

```python
tfe.setEigs(1, 1.1)
```

```python
tfe.dpsi_dl()
```

```python
tfe.d2psi_dl2()
```

```python
spacing = np.linspace(0.9, 1.5**0.5, 200)
X, Y = np.meshgrid(spacing, spacing)

E = Y.copy()
E.fill(np.nan)
dE_dl1 = E.copy()
dE_dl2 = E.copy()

for j, x in enumerate(spacing):
    for i, y in enumerate(spacing):
        if (x < y): continue
        tfe.setEigs(x*x, y*y)
        E[i, j] = tfe.psi()
        #E[i, j] = x*x + y*y + 1 / (x*x * y*y) - 3.0
        if E[i, j] > 2000: E[i, j] = np.nan
        dE_dl1[i, j], dE_dl2[i, j] = tfe.dpsi_dl()
```

```python
from matplotlib import pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import warnings; warnings.simplefilter('ignore')
fig = plt.figure(figsize=(24,8))
data = [E, dE_dl1, dE_dl2]
for i, Z in enumerate(data):
    ax = fig.add_subplot(1, 3, i + 1, projection='3d')
    ax.view_init(elev=30, azim=240)
    ax.plot_surface(X, Y, Z)
plt.tight_layout()
plt.show()
```
