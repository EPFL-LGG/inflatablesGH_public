```python
import sys
sys.path.append('..')
import metric_fit, parametrization, numpy as np, mesh
from tri_mesh_viewer import TriMeshViewer
```

```python
refConfig = [[0, 1, 0], [1, 0, 0], [0, 0, 0], [1, 1, 0]]
deformedConfig = lambda theta: np.array([[0, 0, -0.5], [0, 0, 0.5], [-np.cos(theta), -np.sin(theta), 0], [1, 0, 0]])

def hingeVisualization(config):
    hingeViewer = TriMeshViewer(mesh.Mesh(config, [[2, 1, 0], [0, 1, 3]]))
    hingeViewer.showWireframe(True)
    return hingeViewer

hingeVisualization(deformedConfig(np.pi / 4)).show()
```

```python
he = metric_fit.HingeEnergy(*refConfig)
th = np.pi / 4
he.setDeformedConfiguration(*deformedConfig(th))
```

```python
delta_p = np.random.uniform(low=-1, high=1, size=(3, 4))
#delta_p[:, 0] = 0
#delta_p[:, 2] = 0
#delta_p[:, 3] = 0
```

```python
he.setDeformedConfiguration(*(deformedConfig(th) + 1e-5 * np.random.uniform(low=-1, high=1, size=(4, 3))))
```

```python
dc = deformedConfig(th)
def thetaAt(x):
    he.setDeformedConfiguration(*x)
    return he.theta
eps = 1e-6
fd_delta_theta = (thetaAt(dc + eps * delta_p.transpose()) - thetaAt(dc - eps * delta_p.transpose())) / (2 * eps)
an_delta_theta = np.dot(he.gradTheta().ravel(), delta_p.ravel())
(fd_delta_theta, an_delta_theta)
```

```python
np.set_printoptions(linewidth=300)
```

```python
dc = deformedConfig(th)
def gradAt(x):
    he.setDeformedConfiguration(*x)
    return he.gradTheta().transpose().ravel()
eps = 1e-9
fd_delta_grad = (gradAt(dc + eps * delta_p.transpose()) - gradAt(dc - eps * delta_p.transpose())) / (2 * eps)
an_delta_grad = he.hessTheta() @ delta_p.transpose().ravel()
(fd_delta_grad, an_delta_grad)
```

```python
dc = deformedConfig(th)
def energyAt(x):
    he.setDeformedConfiguration(*x)
    return he.energy()
eps = 1e-6
fd_delta_energy = (energyAt(dc + eps * delta_p.transpose()) - energyAt(dc - eps * delta_p.transpose())) / (2 * eps)
an_delta_energy = np.dot(he.gradient().ravel(), delta_p.ravel())
(fd_delta_energy, an_delta_energy)
```

```python
dc = deformedConfig(th)
def gradAt(x):
    he.setDeformedConfiguration(*x)
    return he.gradient().transpose().ravel()
eps = 1e-9
fd_delta_grad = (gradAt(dc + eps * delta_p.transpose()) - gradAt(dc - eps * delta_p.transpose())) / (2 * eps)
an_delta_grad = he.hessian() @ delta_p.transpose().ravel()
(fd_delta_grad, an_delta_grad)
```
