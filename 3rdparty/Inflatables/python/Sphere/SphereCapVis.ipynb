```python
import numpy as np
def concatVisGeometries(A, B):
    return (np.vstack([A[0], B[0]]), # Stacked V
            np.vstack([A[1], B[1] + len(A[0])]), # Stacked F
            np.vstack([A[2], B[2]])) # Stacked N
def concatWithColors(A, colorA, B, colorB):
    return concatVisGeometries(A, B), np.vstack([np.tile(colorA, [len(A[0]), 1]), np.tile(colorB, [len(B[0]), 1])])
```

```python
import glob
```

```python
for cap_dir in glob.glob('cap_*'):
    cap_opt = utils.load(f'{cap_dir}/opt.pkl.gz')

    tsf = cap_opt.rso.targetSurfaceFitter()
    target_surf = mesh.Mesh(tsf.targetSurfaceV, tsf.targetSurfaceF)
    isheet = cap_opt.rso.targetAttractedInflation().sheet()

    v = OViewer(target_surf, width=2048, height=2048)

    visGeom, C = concatWithColors(isheet.visualizationGeometry(), [1, 0, 0, 0.33], target_surf.visualizationGeometry(), [0.85, 0.85, 0.85, 0.33])
    v.renderer.setMesh(*visGeom, C)

    v.renderer.render()
    v.renderer.save(f'{cap_dir}/deviation_top.png')
    v.setCameraParams(((3.748644842161836, 3.142507784572672, 1.1726674221157354),
                       (-0.26228557780135153, -0.13847218021685662, 0.9550035240685767),
                       (0.09496113817230088, -0.0030604333809495553, -0.2868897370315449)))
    v.renderer.render()
    v.renderer.save(f'{cap_dir}/deviation_side.png')
```
