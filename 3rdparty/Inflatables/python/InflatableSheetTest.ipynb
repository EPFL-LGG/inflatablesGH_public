```python
import inflation, numpy as np, importlib, fd_validation
```

```python
# m = inflation.Mesh('../examples/half_sphere.msh')
# m = inflation.Mesh('../examples/octagon.obj')
m = inflation.Mesh('../examples/single_tri.obj')
isheet = inflation.InflatableSheet(m)
```

```python
isheet.rigidMotionPinVars, isheet.getVars()[isheet.rigidMotionPinVars]
```

```python
Pressure = inflation.InflatableSheet.EnergyType.Pressure
Elastic  = inflation.InflatableSheet.EnergyType.Elastic
Full     = inflation.InflatableSheet.EnergyType.Full
isheet.pressure = 1
```

```python
perturb = np.random.uniform(low=-1.0, high=1.0, size=isheet.numVars())
perturb2 = np.random.uniform(low=-1.0, high=1.0, size=isheet.numVars())
```

```python
fd_validation.validateGrad(isheet, xeval=isheet.getVars() + 1e-2 * perturb,
                           etype=Elastic, fd_eps=1e-6)
```

```python
xorig = isheet.getVars()
xeval = isheet.getVars() + 1e-2 * perturb
```

```python
isheet.setVars(xeval)
```

```python
isheet.setUseTensionFieldEnergy(True)
```

```python
isheet.cauchyGreenDeformationTensors()
```

```python
isheet.energy(Elastic)
```

```python
isheet.tensionStateHistogram()
```

```python
H_err = fd_validation.validateHessian(isheet, fd_eps=1e-7, xeval=xeval, perturb=perturb2, etype=Elastic)
fd_delta_grad, an_delta_grad = H_err[1:]
H_err
```

```python
worst = np.argmax(np.abs(an_delta_grad - fd_delta_grad))
(an_delta_grad[max(worst-2, 0):worst+2], fd_delta_grad[max(worst-2, 0):worst+2])
```
