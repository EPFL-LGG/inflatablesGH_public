```python
import inflation, numpy as np, importlib
from numpy.linalg import norm
m = inflation.Mesh('../examples/half_sphere.msh')
# m = inflation.Mesh('../examples/octagon_highres.msh')
# m = inflation.Mesh('../examples/octagon.obj')
# m = inflation.Mesh('../examples/single_tri.obj')
isheet = inflation.InflatableSheet(m)
```

```python
Pressure = inflation.InflatableSheet.EnergyType.Pressure
Elastic  = inflation.InflatableSheet.EnergyType.Elastic
Full     = inflation.InflatableSheet.EnergyType.Full
opts = inflation.NewtonOptimizerOptions()
opts.useIdentityMetric = True
opts.beta = 1e-4
opts.niter = 5000
opts.gradTol = 1e-8
```

```python
isheet.setUseTensionFieldEnergy(False)
inflation.benchmark_reset()
for step, pressure in enumerate(np.linspace(0.0, 3.5, 20)):
    isheet.pressure = pressure
    cr = inflation.inflation_newton(isheet, isheet.rigidMotionPinVars, opts)
    print(pressure, isheet.energy(), isheet.energy(Pressure), isheet.energy(Elastic))
    print(isheet.getVars()[isheet.rigidMotionPinVars])
    isheet.writeDebugMesh('octagon_pillow/inflation_{}.msh'.format(step))
inflation.benchmark_report()
```

```python
inflation.benchmark_reset()
isheet.setUseTensionFieldEnergy(True)
for step, pressure in enumerate(np.linspace(0.0, 3.5, 20)):
    isheet.pressure = pressure
    cr = inflation.inflation_newton(isheet, isheet.rigidMotionPinVars, opts)
    isheet.writeDebugMesh('octagon_pillow_hires/inflation_{}.msh'.format(step))
inflation.benchmark_report()
```

```python
inflation.benchmark_reset()
isheet.setUseTensionFieldEnergy(True)
for step, pressure in enumerate(np.linspace(0.0, 0.1, 3)):
    isheet.pressure = pressure
    cr = inflation.inflation_newton(isheet, isheet.rigidMotionPinVars, opts)
    isheet.writeDebugMesh('half_sphere_tension_field_inflation/step_{}.msh'.format(step))
inflation.benchmark_report()
```

```python
inflation.benchmark_reset()
isheet.setUseTensionFieldEnergy(False)
for step, pressure in enumerate(np.linspace(0.0, 0.1, 3)):
    isheet.pressure = pressure
    cr = inflation.inflation_newton(isheet, isheet.rigidMotionPinVars, opts)
    isheet.writeDebugMesh('half_sphere_neo_hookean_inflation/step_{}.msh'.format(step))
inflation.benchmark_report()
```
