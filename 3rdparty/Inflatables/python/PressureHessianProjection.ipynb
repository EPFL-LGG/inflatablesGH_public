```python
import numpy as np
from numpy.linalg import norm
```

```python
def cross_matrix(v):
    return np.array([[0, -v[2], v[1]], [v[2], 0, -v[0]], [-v[1], v[0], 0]])
np.zeros((3,3))
```

```python
class pressurePotential:
    def __init__(self, V):
        self.V = V
        e0 = V[:, 1] - V[:, 0]
        e1 = V[:, 2] - V[:, 0]
        print(V)
        print(e0)
        print(e1)
        self.nA = 0.5 * np.cross(e0, e1)
        self.A = norm(self.nA)
        self.n = self.nA / self.A
    def v(self, i): return self.V[:, i]
    def gradient(self):
        return -self.nA / 3.0
    def hessian(self):
        Z = np.zeros((3,3))
        return np.block(
            [[                       Z, -cross_matrix(self.v(2)),  cross_matrix(self.v(1))],
             [ cross_matrix(self.v(2)),                        Z, -cross_matrix(self.v(0))],
             [-cross_matrix(self.v(1)),  cross_matrix(self.v(0)),                        Z]])
```
