```python
import inflation, numpy as np, importlib, fd_validation
```

```python
m, fuseMarkers = inflation.triangulate_channel_walls([[0, 0], [1, 0], [1, 1], [0, 1], [0.5, 0.05], [0.5, 0.95]], [(0, 1), (1, 2), (2, 3), (3, 0), (4, 5)], 0.001)
```

```python
import visualization
importlib.reload(visualization)
visualization.plot_2d_mesh(m, pointList=np.where(np.array(fuseMarkers) == 1)[0], width=12, height=12)
```

```python
isheet = inflation.InflatableSheet(m, np.array(fuseMarkers) != 0)
```

```python
opts = inflation.NewtonOptimizerOptions()
opts.useIdentityMetric = True
opts.beta = 1e-4
opts.niter = 5000
opts.gradTol = 1e-8
```

```python
inflation.benchmark_reset()
isheet.setUseTensionFieldEnergy(True)
for step, pressure in enumerate(np.linspace(0.0, 0.001, 2)):
    isheet.pressure = pressure
    cr = inflation.inflation_newton(isheet, isheet.rigidMotionPinVars, opts)
    isheet.writeDebugMesh('square_test/inflation_tf_{}.msh'.format(step))
inflation.benchmark_report()
```
