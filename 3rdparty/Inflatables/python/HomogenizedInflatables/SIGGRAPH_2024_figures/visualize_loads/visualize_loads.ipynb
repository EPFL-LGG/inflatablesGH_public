```python
import numpy as np
```

```python
old_pattern_side_load_weight = np.cumsum([33,574,505,436])
```

```python
old_pattern_side_deflection = np.cumsum([0, 5, 13, 31])
```

```python
new_pattern_side_load_weight = np.cumsum([33, 574, 505, 436, 368, 368, 299])
```

```python
new_pattern_side_deflection = np.cumsum([0, 2.8, 2.7, 3, 3.6, 4, 4.7])
```

```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(figsize = (12, 7))
plt.plot(old_pattern_side_load_weight, old_pattern_side_deflection, label = "Parallel tubes - [Panetta et al. 2021]")
plt.plot(new_pattern_side_load_weight, new_pattern_side_deflection, label = "Cosine curves - Ours")
ax.legend(loc = 'upper left')


fig.savefig("load_displacement_curves.svg", dpi = 300)
```
