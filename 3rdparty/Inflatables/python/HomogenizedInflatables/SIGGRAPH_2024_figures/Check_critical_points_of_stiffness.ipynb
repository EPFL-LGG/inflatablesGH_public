```python
import numpy as np
```

```python
x = np.linspace(0, np.pi, 1000)
```

```python
q = [1, 20, 20, 20, 1]
```

```python
q
```

```python
from ipywidgets import interact, interactive, fixed, interact_manual
from ipywidgets import interact, widgets
```

```python
def plot_curve(q1, q2, q3, q4, q5):
    q = [q1, q2, q3, q4, q5]
    y = np.zeros(len(x))
    for i in range(5):
        y += q[i] * np.cos(x) ** (4 - i) * np.sin(x) ** i

    import matplotlib.pyplot as plt
    fig = plt.figure()
    ax = fig.add_subplot(1, 1, 1)

    # line, = ax.plot(a, color='blue', lw=2)

    # ax.set_yscale('log')

    plt.plot(x, y)
```

```python
q_slider = []
for i in range(5):
    q_slider.append(widgets.FloatSlider(min=0, max=1000, step=1, value=0))
w = interact(plot_curve, q1 = q_slider[0], q2 = q_slider[1], q3 = q_slider[2],q4 = q_slider[3], q5 = q_slider[4]);
```
