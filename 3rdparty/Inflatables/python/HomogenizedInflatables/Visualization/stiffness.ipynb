```python
import json
import sys; sys.path.append("../..")
import MeshFEM
```

```python
labels = np.arange(51)
length_values = []
for label in labels:
    with open('../data/ZigZag_Line/ZigZag_Line{}.json'.format(label), 'r') as f:
        data = json.load(f)
    fusedVertices = data['FusedVertices']
    fusedVertices = [True if vx == 1 else False for vx in fusedVertices]
    V = data['Vertices']
    F = data['Faces']
    m = MeshFEM.Mesh(V, F)
    for vx in m.boundaryVertices():
        fusedVertices[vx] = False
    fuse_pos = np.array(V)[np.where(np.array(fusedVertices) == True)[0], :]
    length_values.append([min(fuse_pos[:, 0]), 5, 10])
length_values = np.array(length_values)
```
