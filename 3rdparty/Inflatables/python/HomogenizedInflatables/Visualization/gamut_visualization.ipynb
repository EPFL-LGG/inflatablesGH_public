```python
import matplotlib.pyplot as plt
import numpy as np
import visualize_stiffness
```

```python
data_info = []
```

```python

stiffness_path = "../experiments/output/grid_dots/2023_04_26_19_57/"
name = "grid_dots"

iteration_tags = range(50)
data_info.append([stiffness_path, name, iteration_tags])
```

```python

stiffness_path = "../experiments/output/three_star/2023_04_26_19_43/"
name = "three_star"

radius = np.linspace(1, 2.4, 15)
angles = np.linspace(0, 15, 16)

iteration_tags = ["{:.1f}_{:.2f}".format(r, angle) for r in radius for angle in angles]
data_info.append([stiffness_path, name, iteration_tags])
```

```python

stiffness_path = "../experiments/output/zigzag_line_10_to_90/2023_04_21_18_30"
name = "zigzag_line_10_to_90"

iteration_tags = np.arange(50)
data_info.append([stiffness_path, name, iteration_tags])
```

```python
stiffness_path = "../experiments/output/zero_area_zigzag_dash/2023_04_20_15_14/"
name = "zero_area_zigzag_dash"

iteration_tags = ['%.2f'%angle for angle in np.linspace(0, 90, 46)]
data_info.append([stiffness_path, name, iteration_tags])
```

```python
stiffness_path = "../experiments/output/cosine_curve_no_mirror/2023_04_23_17_35//"
name = "cosine_curve_no_mirror"

iteration_tags = ['%.2f'%amp for amp in np.linspace(0, 3, 61)]
data_info.append([stiffness_path, name, iteration_tags])
```

```python
stiffness_path = "../experiments/output/mirror_dash/2023_04_24_10_09//"
name = "mirror_dash"

angles = np.linspace(0, 90, 46)
radiuses = np.linspace(1, 2.4, 15)

iteration_tags = [ "{}_{:.2f}".format(angle, r) for angle in angles for r in radiuses]
data_info.append([stiffness_path, name, iteration_tags])
```

```python
data_collection = [visualize_stiffness.plot_all_data(None, *data_info[i]) for i in range(len(data_info))]
```

```python
fig, ax = plt.subplots()
# ax.set_ylim(ymin=0.5)
# ax.set_xlim(xmin=0.5)
ax.set_aspect("equal")

for i in range(len(data_info)):
    plt.scatter(np.array(data_collection[i][-1])[:, 0], np.array(data_collection[i][-1])[:, 1], label = data_info[i][1], s = 50, alpha = 0.3)



fig.set_size_inches(12, 12)
ax.title.set_text("Scale factors")
plt.xlabel("Min scale factors")
plt.ylabel("Max scale factors")

# Plot x = y line
lims = [
    np.min([ax.get_xlim(), ax.get_ylim()]),  # min of both axes
    np.max([ax.get_xlim(), ax.get_ylim()]),  # max of both axes
]

# now plot both limits against eachother
ax.plot(lims, lims, 'k-', alpha=0.75, zorder=0)
ax.set_aspect('equal')
ax.set_xlim(lims)
ax.set_ylim(lims)
ax.legend()
fig.tight_layout()
plt.savefig('scale_factors.png', dpi = 300)
```
