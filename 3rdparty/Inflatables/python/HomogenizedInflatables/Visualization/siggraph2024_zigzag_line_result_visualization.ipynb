```python
parallel_tube_scale_factors = np.ones((len(scale_factor_data), 2))
parallel_tube_scale_factors[:, 0] *= scale_factor_data[-1][0]
parallel_tube_scale_factors[:, 1] *= scale_factor_data[-1][1]
```

```python
angles = np.linspace(10, 90, 50)[10:]
angles *= np.pi / 180
```

```python
analytical_scale_factors = visualize_stiffness.analytical_zigzag_scale_factors_angles(angles, parallel_tube_scale_factors = parallel_tube_scale_factors, total_area = None)


analytical_scale_factors = np.array(analytical_scale_factors[:2])
analytical_scale_factors = list(analytical_scale_factors.transpose())
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
```

```python
fig, ax = plt.subplots()

    
violin_parts = plt.violinplot(scale_factor_data, positions = np.arange(len(scale_factor_data)))
for pc in violin_parts['bodies']:
    pc.set_facecolor('white')
    
violin_parts = plt.violinplot(analytical_scale_factors, positions = np.arange(len(scale_factor_data)), )
for pc in violin_parts['bodies']:
    pc.set_facecolor('white')



ax.legend((mpatches.Patch(color="tab:orange"), mpatches.Patch(color="tab:blue")), ('Geometric abstraction', 'Simulation-based homogenization'), loc = "lower center")


fig.set_size_inches(12, 4)

ax.title.set_text("Max and min scale factors")

fig.tight_layout()

fig.canvas.draw()
labels = [(item.get_text()) for item in ax.get_xticklabels()]
# select_theta = [int(x) for x in labels[1:-1]]
# print(select_theta)
# print(labels)
def get_angle(i):
    return 80 / 49 * i + angles[10]
# labels[1:-1] = np.round([get_angle(x) for x in select_theta], 2)
# ax.set_xticklabels(labels)

select_theta = np.array([0, 9, 19, 29, 39])
angles = np.linspace(10, 90, 50)[10:]

ax.set_xticklabels(np.round([angles[x] for x in select_theta], 2))
ax.set_xticks(select_theta)


plt.savefig('scale_factors_comparison_{}.svg'.format(name), dpi = 300)

plt.show()
```
