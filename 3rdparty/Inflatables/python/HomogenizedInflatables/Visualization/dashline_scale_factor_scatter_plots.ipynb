```python
import matplotlib.cm as cm

fig,axs=plt.subplots(1, 3, figsize = (15, 4))
axs[0].scatter(data[:, 0], data[:, 1], c = data[:, 2], cmap=cm.Set1, vmin = 0, vmax = 1)
axs[0].set_aspect('equal', 'box')
axs[0].title.set_text("Convergence")

axs[1].scatter(data[:, 0], data[:, 1], c = data[:, 3], cmap=cm.cool, vmin = min_scale, vmax = max_scale)
axs[1].set_aspect('equal', 'box')
axs[1].title.set_text("Min scale factors")


axs[2].scatter(data[:, 0], data[:, 1], c = data[:, 4], cmap=cm.cool, vmin = min_scale, vmax = max_scale)
axs[2].set_aspect('equal', 'box')
axs[2].title.set_text("Max scale factors")


fig.tight_layout()
plt.savefig('dashline_20_0_1_result.png', dpi = 300)
```
