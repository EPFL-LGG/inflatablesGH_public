```python
data_info = []
```

```python
def visualize_data(select):
    stiffness_path = "../experiments/output/spiral/2023_05_17_12_56//"
    name = "spiral"

    start_angles = np.linspace(0, 180, 21)
    alphas = np.linspace(3, 89, 21)

    iteration_tags = ["{:.2f}_{:.2f}".format(start, a) for start in start_angles for a in alphas]
    data_info.append([stiffness_path, name, iteration_tags, name + " " + str(select)])

    len(iteration_tags)

    fig, ax = plt.subplots()
    # ax.set_ylim(ymin=0.5)
    # ax.set_xlim(xmin=0.5)
    ax.set_aspect("equal")

    for i in range(len(data_info)):

        used_tags = visualize_stiffness.get_used_tags(*data_info[i][:3])

        # used_tags

        # data = visualize_stiffness.plot_all_data(None, data_info[i][0], data_info[i][1], used_tags)

        scale_factor_matrix_data = visualize_stiffness.get_scale_factor_matrix(data_info[i][0], data_info[i][1], used_tags)

        import numpy.linalg as la


        x_scale_factor = [1./ la.norm(mat @ np.array([1, 0])) for mat in scale_factor_matrix_data]
        y_scale_factor = [1./ la.norm(mat @ np.array([0, 1])) for mat in scale_factor_matrix_data]

        plt.scatter(x_scale_factor, y_scale_factor, label = data_info[i][3], s = 50, alpha = 0.3)
        # plt.scatter(y_scale_factor, x_scale_factor, label = data_info[i][1], s = 50, alpha = 0.3)



    fig.set_size_inches(12, 12)
    ax.title.set_text("Scale factors")
    plt.xlabel("x scale factors")
    plt.ylabel("y scale factors")

    # Plot x = y line
    lims = [
        np.min([ax.get_xlim(), ax.get_ylim()]),  # min of both axes
        np.max([ax.get_xlim(), ax.get_ylim()]),  # max of both axes
    ]

    # now plot both limits against eachother
    ax.plot(lims, lims, 'k-', alpha=0.75, zorder=0)
    ax.set_aspect('equal')
    ax.set_xlim(lims)
    ax.set_ylim(lims)
    ax.legend()
    fig.tight_layout()
    # plt.savefig('scale_factor_values_{}.png'.format(name), dpi = 300)
```
